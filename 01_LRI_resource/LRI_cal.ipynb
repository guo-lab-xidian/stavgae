```python
import pandas as pd
import numpy as np

# Step 1: 读取数据
L_expression = pd.read_csv('D:/yxhu/GSE197177/preprocess3/LRI/L_fb2.3_细胞类型表达值.csv', index_col=0)  # 配体基因表达矩阵
R_expression = pd.read_csv('D:/yxhu/GSE197177/preprocess3/LRI/R_fb2.3_细胞类型表达值.csv', index_col=0)  # 受体基因表达矩阵
LRI_data = pd.read_csv('D:/yxhu/GSE197177/preprocess3/LRI/cfb2.3_LRI.csv')  # 配体-受体信息
```

```python
# Step 2: 初始化结果存储
L_final = pd.DataFrame(index=LRI_data.index, columns=L_expression.columns)  # 多亚基配体结果
R_final = pd.DataFrame(index=LRI_data.index, columns=R_expression.columns)  # 多亚基受体结果
```

```python
# Step 3: 逐行计算多亚基表达值
for idx, row in LRI_data.iterrows():
    # 获取配体和受体基因
    ligands = [row[f'ligand{i+1}'] for i in range(3) if pd.notna(row[f'ligand{i+1}'])]
    receptors = [row[f'receptor{i+1}'] for i in range(3) if pd.notna(row[f'receptor{i+1}'])]
    
    # 计算多亚基配体表达值
    if ligands:
        ligand_values = L_expression.loc[ligands].prod(axis=0) ** (1 / len(ligands))
    else:
        ligand_values = pd.Series(0, index=L_expression.columns)
    
    # 计算多亚基受体表达值
    if receptors:
        receptor_values = R_expression.loc[receptors].prod(axis=0) ** (1 / len(receptors))
    else:
        receptor_values = pd.Series(0, index=R_expression.columns)
    
    # 存储结果
    L_final.loc[idx] = ligand_values
    R_final.loc[idx] = receptor_values

# Step 4: 输出结果
L_final.to_csv('L_final_expression.csv')  # 保存多亚基配体表达结果
R_final.to_csv('R_final_expression.csv')  # 保存多亚基受体表达结果

print("多亚基配体和受体表达值已计算并保存为 'L_final_expression.csv' 和 'R_final_expression.csv'")
```
